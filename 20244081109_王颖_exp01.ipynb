## 实验2：函数、条件、循环与递归

**任务1**：`score_level(score)`：90~100→“优秀”，80~不足90→“良好”，60~不足80→“及格”，0~不足60→“不及格”；超出 0~100 抛 `ValueError`。


````python
# 任务1：成绩等级函数
def score_level(score):
    """根据分数返回等级，超出 0~100 范围时抛出 ValueError"""
    if not (0 <= score <= 100):
        raise ValueError(f"分数 {score} 超出 0~100 范围")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

print(score_level(95), score_level(85), score_level(60), score_level(59))
try:
    score_level(120)
except ValueError as e:
    print("越界抛错:", e)
````

**任务2**：分别用 `for` 和 `while` 计算 1 到 100 的整数和，比较结果。


````python
# 任务2：for 与 while 计算 1~100 的和
sum_for = 0
for i in range(1, 101):
    sum_for += i

sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

print("for 结果:", sum_for)
print("while 结果:", sum_while)
print("两种结果一致:", sum_for == sum_while)
````

**任务3**：编写递归版和迭代版阶乘函数，约定 $0! = 1$，输入为非负整数。比较 n = 0、1、5、10 时的结果。


````python
# 任务3：递归版与迭代版阶乘
def factorial_recursive(n):
    """递归实现：n! = n * (n-1)!，基准情形 0! = 1"""
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    """迭代实现：从 2 一直乘到 n"""
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

for n in (0, 1, 5, 10):
    print(f"n={n}: 递归={factorial_recursive(n)}  迭代={factorial_iterative(n)}")
````

**任务4**：`summary(values)` 接收非空数值列表，以元组返回最小值、最大值和平均值，并通过元组解包接收结果。


````python
# 任务4：summary 返回 (最小值, 最大值, 平均值)
def summary(values):
    """接收非空数值列表，以元组返回最小值、最大值、平均值"""
    return min(values), max(values), sum(values) / len(values)

mn, mx, avg = summary([4, 8, 2, 6, 9])   # 元组解包
print("最小值:", mn, "最大值:", mx, "平均值:", avg)
````

**任务5**：`sum_all(*args)` 返回所有参数之和；分别测试无参数、单个参数和多个参数的情况。


````python
# 任务5：sum_all(*args)
def sum_all(*args):
    """*args 收集任意数量的位置参数，返回总和"""
    return sum(args)

print("无参数:", sum_all())
print("单个参数:", sum_all(5))
print("多个参数:", sum_all(1, 2, 3, 4, 5))
````

## 实验4：文件读写、异常处理与模块

说明：在 Notebook 同级目录创建 `exp01_data` 文件夹，保存 UTF-8 文本文件 `students.txt`。


````python
# 任务1：查看工作目录、构造相对路径、创建数据目录
import os

print("当前工作目录:", os.getcwd())

data_dir = os.path.join(os.getcwd(), "exp01_data")  # 构造相对路径
os.makedirs(data_dir, exist_ok=True)                 # 已存在时不报错
print("数据目录:", data_dir)
````

````python
# 任务2：用 with open 写入上述数据，再读取文件
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(os.path.join(data_dir, "students.txt"), "w", encoding="utf-8") as f:
    f.write(content)

with open(os.path.join(data_dir, "students.txt"), "r", encoding="utf-8") as f:
    lines = f.readlines()
print(lines)
````

````python
# 任务3：parse_student(line) —— 解析一行学生记录
def parse_student(line):
    """解析 '姓名,成绩' 行；字段数不对、姓名为空、成绩非整数或超范围时抛 ValueError"""
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不为 2")
    name, score_str = parts
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩无法转为整数: {score_str}")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return name, score

print(parse_student("Alice,85"))
try:
    parse_student("Grace,abc")
except ValueError as e:
    print("无效记录报错:", e)
````

````python
# 任务4：按行处理文件，捕获无效记录的异常，记录行号与原因，继续处理后续记录
valid_records = []    # 有效记录：字典组成的列表
invalid_log = []      # 无效记录：(行号, 错误原因)

with open(os.path.join(data_dir, "students.txt"), "r", encoding="utf-8") as f:
    raw_lines = f.readlines()

for lineno, line in enumerate(raw_lines, start=1):
    if lineno == 1 or not line.strip():   # 跳过表头与空行
        continue
    try:
        name, score = parse_student(line)
        valid_records.append({"name": name, "score": score})
    except ValueError as e:
        invalid_log.append((lineno, str(e)))

print("有效记录:", valid_records)
print("无效记录(行号, 原因):", invalid_log)
````

````python
# 任务5：统计并写入 report.txt，分别演示 w 与 a 模式
n_valid = len(valid_records)
n_invalid = len(invalid_log)
avg = sum(r["score"] for r in valid_records) / n_valid
pass_rate = len([r for r in valid_records if r["score"] >= 60]) / n_valid

report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:   # w 模式：覆盖写入，清空已有内容
    f.write(f"有效记录数: {n_valid}\n")
    f.write(f"无效记录数: {n_invalid}\n")
    f.write(f"平均分: {avg:.2f}\n")
    f.write(f"及格率: {pass_rate:.1%}\n")

with open(report_path, "a", encoding="utf-8") as f:   # a 模式：追加写入，保留已有内容
    f.write("（这是用 a 模式追加的一行，说明 a 不清空已有内容）\n")

with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
````

## 实验5：类、继承与对象复制

**任务1**：定义 `Student` 类，`__init__()` 初始化 `name` 和 `score`，实现 `is_passed()` 和 `__str__()`。


````python
# 任务1：定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩 >= 60 为及格"""
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"
````

**任务2**：为实验4中的每条有效记录创建一个 Student 对象，输出对象信息和及格状态。


````python
# 任务2：为每条有效记录创建 Student 对象
students = [Student(r["name"], r["score"]) for r in valid_records]
for stu in students:
    print(stu, "| 及格" if stu.is_passed() else "| 不及格")
````

**任务3**：定义继承自 Student 的 `GraduateStudent`，增加 `research_topic` 属性，用 `super().__init__()` 初始化父类属性，并重写 `__str__()`。


````python
# 任务3：GraduateStudent 继承 Student
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)   # 调用父类初始化方法
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic})"

g = GraduateStudent("Carol", 92, "自然语言处理")
print(g)
````

**任务4**：构造包含嵌套列表的对象（如 `{"name": "Alice", "scores": [85, 90]}`），分别进行直接赋值、`copy.copy()` 和 `copy.deepcopy()`，修改内部列表后比较结果。每次比较都从新建的原始对象开始。


````python
# 任务4：直接赋值 / 浅拷贝 / 深拷贝（每次从新建的原始对象开始）
import copy

# (a) 直接赋值：新变量只是原对象的另一个名字，共享全部内容
original = {"name": "Alice", "scores": [85, 90]}
a = original
a["scores"].append(95)
print("直接赋值后 → 原对象:", original)

# (b) 浅拷贝：复制外层字典，但内部嵌套列表仍共享
original = {"name": "Alice", "scores": [85, 90]}
b = copy.copy(original)
b["scores"].append(95)
print("浅拷贝后   → 原对象:", original)

# (c) 深拷贝：完全独立的副本，嵌套列表也独立
original = {"name": "Alice", "scores": [85, 90]}
c = copy.deepcopy(original)
c["scores"].append(95)
print("深拷贝后   → 原对象:", original)
````
