## 热身：定义简单函数

1. `say_hi()`：不接受参数，调用时打印 `"Hello, world!"`
2. `goat_pad(s)`：接受一个字符串，打印时前后都加上 `"goat"`
3. `print_n(s, n)`：把字符串 `s` 打印 `n` 次，每次单独一行


```python
def say_hi():
    """不接受参数，调用时打印 Hello, world!"""
    print("Hello, world!")


def goat_pad(s):
    """接受一个字符串参数，打印时前后都加上 goat"""
    print("goat" + s + "goat")


def print_n(s, n):
    """把字符串 s 打印 n 次，每次单独一行"""
    for _ in range(n):
        print(s)
```

## 欧几里得算法（最大公约数 gcd）

同时能整除 m 和 n 的最大整数称为它们的最大公约数。

算法思路：`gcd(m, n) = gcd(n, m % n)`，不断用“余数”替换“较小的数”，直到余数为 0。


```python
def gcd(m, n):
    """返回两个整数的最大公约数（欧几里得算法）"""
    while n != 0:
        m, n = n, m % n
    return m
```

```python
# 测试三组数字
print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
# 观察：参数为负数时函数的行为
print(gcd(-12, 8))
print(gcd(12, -8))
print(gcd(-12, -8))
```

### 负数行为分析（思考题）

- Python 的 `%` 运算结果符号与除数一致，所以算法仍能“算完”，返回的是一个正数。
- 但数学上最大公约数通常只对**非负整数**定义。例如 `gcd(-12, 8)` 返回 4，相当于 `|−12|` 与 8 的公约数。
- 结论：对负数输入，结果虽然“有值”，但含义不完全符合数学定义。规范做法是先在函数里取绝对值 `abs()`，或保证调用时输入非负。


## 近似欧拉数 e

公式(1)（级数）：$$e = \sum_{k=0}^{\infty} \frac{1}{k!} = 1 + \frac{1}{1!} + \frac{1}{2!} + \frac{1}{3!} + \cdots$$，约定 $0! = 1$

公式(2)（极限）：$$e = \lim_{x \to \infty} \left(1 + \frac{1}{x}\right)^x$$

要求实现 4 个函数：`euler_limit` / `euler_infinite_sum` / `euler_approx` / `print_euler_sum_table`


```python
import math

def euler_limit(n):
    """用极限公式 (1 + 1/n)**n 近似 e，取 x = n"""
    return (1 + 1 / n) ** n
```

```python
def euler_infinite_sum(n):
    """返回公式(1)前 n 项之和（n 为 1-索引，项从 k=0 开始）
    euler_infinite_sum(1) -> 1；euler_infinite_sum(0) -> 0"""
    total = 0.0
    fact = 1          # 保存当前 k!，初始 0! = 1
    for k in range(n):
        if k > 0:
            fact *= k # k! = (k-1)! * k
        total += 1 / fact
    return total
```

```python
def euler_approx(epsilon):
    """用级数不断累加，直到与真实 e 的差小于 epsilon，返回近似值"""
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(true_e - total) >= epsilon:
        if k > 0:
            fact *= k
        total += 1 / fact
        k += 1
    return total
```

```python
def print_euler_sum_table(n):
    """打印 k = 1..n 对应的 euler_infinite_sum(k)，每行一个"""
    for k in range(1, n + 1):
        print(k, euler_infinite_sum(k))
```

```python
# 测试
print("limit(100):", euler_limit(100))
print("infinite_sum(4):", euler_infinite_sum(4))   # 应为 1+1+1/2+1/6 ≈ 2.6667
print("infinite_sum(10):", euler_infinite_sum(10))
print("approx(0.0001):", euler_approx(0.0001))
print("真实 e:", math.exp(1))
print("--- 求和表(1..5) ---")
print_euler_sum_table(5)
```

## 测试整数属性

- 正整数 n 是 2 的幂：$n = 2^p$（$p \ge 0$），如 1、2、4、8、16 …
- 只用除法和取模（`%`）实现，不能用 `math.sqrt`
- 推广：`is_power(b, n)` 判断 n 是否为 b 的幂


```python
def is_power_of_2(n):
    """判断正整数 n 是否是 2 的幂：不断整除 2，最后应得到 1 且中途无余数"""
    while n % 2 == 0:
        n = n // 2
    return n == 1
```

```python
def is_power(b, n):
    """判断 n 是否是 b 的幂（b、n 均为正整数）"""
    if n == 1:            # 1 = b^0，任何底数的 0 次幂都是 1
        return True
    while n % b == 0:
        n = n // b
    return n == 1
```

```python
# 测试
print(is_power_of_2(1), is_power_of_2(2), is_power_of_2(8), is_power_of_2(10))
print(is_power(3, 81), is_power(3, 80), is_power(2, 32), is_power(5, 1))
```
