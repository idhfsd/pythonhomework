## 字符串的乐趣
在这个问题中，您将实现⼀些⽤于处理字符串的简单函数。您不需要为这个问题中的任
何函数执⾏错误检查。
1. `is_palindrome(s)`：判断回文。忽略空格和大小写，仅含英文字母、数字、空格。
2. `is_abecedarian(s)`：判断字母是否从左到右按字母顺序排列（允许重复），忽略空格。
3. `remove_vowels(s)`：去掉元音 a、e、i、o、u，返回剩余字符组成的字符串。


```python
# ---------- 1. 回文：忽略空格和大小写 ----------
def is_palindrome(s):
    s = s.replace(" ", "").lower()      # 去空格、统一小写
    return s == s[::-1]                  # 与反转后的字符串比较

# ---------- 2. 字母顺序：忽略空格，允许重复 ----------
def is_abecedarian(s):
    s = s.replace(" ", "")
    # 相邻字符满足 前一个 <= 后一个 即通过；<= 表示允许重复
    return all(s[i] <= s[i + 1] for i in range(len(s) - 1))

# ---------- 3. 去掉元音 a e i o u ----------
def remove_vowels(s):
    vowels = "aeiouAEIOU"               # 大小写都处理
    return "".join(ch for ch in s if ch not in vowels)
```

## 列表的乐趣

1. `list_reverse(lst)`：返回反转后的列表；输入不是列表时引发错误。
2. `is_sorted(seq)`：判断序列是否升序排列；只遍历一次，支持任意可索引、可比较的序列。

```python
# ---------- 1. 反转列表 ----------
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]

# ---------- 2. 判断升序（单次遍历） ----------
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

print(list_reverse([1, 2, 3]))             # [3, 2, 1]
print(is_sorted([1, 2, 3]))                # True
print(is_sorted([1, 3, 2]))                # False
print(is_sorted("abc"))                    # True（字符串也是序列，动态类型）
```

## 作为向量的元组

用长度为 d 的元组表示 d 维向量：

1. `vec_scalar_mult(t, s)`：t 中每个元素乘以标量 s，返回同长度元组；类型不合适抛 TypeError。
2. `vec_inner_product(v1, v2)`：返回内积（浮点数）；非元组抛 TypeError，长度不一致抛 ValueError。
3. `check_valid_mx(mx)`：判断是否为有效矩阵（元组的元组，各内层元组等长，元素均为整数或浮点数）。
4. `mx_vec_mult(mx, v)`：矩阵乘向量，返回数值元组；用 check_valid_mx 校验，不合理抛错误。

```python
# ---------- 1. 标量乘法 ----------
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("t 必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("s 必须是整数或浮点数")
    if not all(isinstance(x, (int, float)) for x in t):
        raise TypeError("t 的元素必须是整数或浮点数")
    return tuple(x * s for x in t)

# ---------- 2. 内积：Σ x_j * y_j ----------
def vec_inner_product(v1, v2):
    if not isinstance(v1, tuple) or not isinstance(v2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(v1) != len(v2):
        raise ValueError("两个向量长度不一致")
    return float(sum(x * y for x, y in zip(v1, v2)))

# ---------- 3. 矩阵合法性检查 ----------
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    if not all(isinstance(row, tuple) for row in mx):
        return False
    n = len(mx[0])
    if not all(len(row) == n for row in mx):
        return False
    return all(isinstance(x, (int, float)) for row in mx for x in row)

# ---------- 4. 矩阵乘向量 ----------
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise TypeError("mx 不是有效矩阵")
    if not isinstance(v, tuple):
        raise TypeError("v 必须是元组")
    if len(mx) == 0:
        raise ValueError("矩阵不能为空")
    if len(v) != len(mx[0]):
        raise ValueError("向量长度与矩阵列数不一致")
    return tuple(sum(x * y for x, y in zip(row, v)) for row in mx)

print(vec_scalar_mult((1, 2, 3), 2))               # (2, 4, 6)
print(vec_inner_product((1, 2, 3), (4, 5, 6)))     # 32.0
print(check_valid_mx(((1, 2), (3, 4))))            # True
print(mx_vec_mult(((1, 2), (3, 4)), (5, 6)))       # (17, 39)
```

## 更多向量的乐趣（面向对象）

1. 定义 `Vector` 类：维度（非负整数）+ 元素（列表/元组）；只给维度时默认全 0 向量；维度或元素非法时抛异常。
2. 维度与元素用**实例属性**（理由见框 13）。
3. 实现比较运算符：坐标完全相同则相等，否则按元素字典序比较；维度不同抛异常。
4. 实现 `dot(other)`：返回内积；类型或维度不对抛异常。
5. 实现乘法：标量左乘/右乘（2*v、v*2）、向量×向量逐元素相乘；维度不一致抛异常。

```python
class Vector:
    def __init__(self, dim, elements=None):
        # 1) 维度检查
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        # 2) 只给维度不给元素 → 默认全 0 向量
        if elements is None:
            elements = [0.0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("元素个数与维度不一致")
        if not all(isinstance(x, (int, float)) for x in elements):
            raise TypeError("元素必须是整数或浮点数")
        # 存为元组，防止外部意外修改
        self.dim = dim
        self.elements = tuple(elements)

    # 维度一致性检查（比较 / 点积 / 乘法共用）
    def _check_dim(self, other):
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能进行该操作")

    # ---------- 比较运算符 ----------
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_dim(other)
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 比较")
        self._check_dim(other)
        return self.elements < other.elements            # 元组自带字典序

    def __le__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 比较")
        self._check_dim(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 比较")
        self._check_dim(other)
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 比较")
        self._check_dim(other)
        return self.elements >= other.elements

    # ---------- 内积 ----------
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector")
        self._check_dim(other)
        return sum(x * y for x, y in zip(self.elements, other.elements))

    # ---------- 乘法：向量×向量（逐元素）/ 向量×标量 ----------
    def __mul__(self, other):
        if isinstance(other, Vector):
            self._check_dim(other)
            return Vector(self.dim, tuple(x * y for x, y in zip(self.elements, other.elements)))
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        raise TypeError("只能与标量或 Vector 相乘")

    # ---------- 标量左乘：2 * v ----------
    def __rmul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, tuple(x * other for x in self.elements))
        raise TypeError("只能与标量或 Vector 相乘")

    # ---------- 便于打印 ----------
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

v1 = Vector(2, (2, 4)); v2 = Vector(2, (2, 5))
print(v1 == Vector(2, (2, 4)), v1 < v2)      # True True
print(v1.dot(v2))                             # 24
print(2 * v1)                                 # Vector(2, (4, 8))
print(v1 * Vector(2, (2, 3)))                 # Vector(2, (4, 12))
#维度与元素为什么用实例属性：类属性（class 内直接赋值）被所有实例共享，而每个 Vector 对象有自己的维度和自己的元素；用实例属性才能保证每个向量独立。若用类属性，创建新向量会覆盖所有向量的数据。
```
